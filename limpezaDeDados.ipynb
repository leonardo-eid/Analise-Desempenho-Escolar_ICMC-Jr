# Resolvendo as inconsistências

A coluna 'student_id' é irrelevante e pode nos atrapalhar no futuro. Assim, eliminamos ela do dataframe

```python
# A função drop() joga a coluna fora
# axis=1 avisa que é uma coluna (se fosse axis=0, seria uma linha)
# inplace=True diz para o pandas salvar a alteração na própria variável 'df'
df.drop('student_id', axis=1, inplace=True)

# Confirmação que o ID sumiu
df.head(3)
```

### Corrigindo pontuações e porcentagens que passam de 100

Algumas pontuações de disciplinas e a taxa de lição de casa (homework_completion_rate) têm anomalias acima de 100. Conforme a orientação dos coordenadores do projeto, vamos varrer essas colunas e excluir qualquer observação (linha) que seja maior que 100.

```python
# Lista das colunas
colunas_verificar = [
    'math_prev_score',
    'science_prev_score',
    'language_prev_score',
    'homework_completion_rate'
]

# O loop passa por cada coluna da lista acima
for coluna in colunas_verificar:
    # A tabela agora será igual a ela mesma, mas SÓ com as linhas onde essa coluna é <= 100
    df = df[df[coluna] <= 100]

# Imprime quantas linhas e colunas sobraram no final
print("Tamanho do banco de dados após a limpeza:")
print(df.shape)
```

### Transformando texto em números

As colunas de nota, escolaridade e nível de ruído são variavéis ordinais, isto é, possuem uma ordem hierárquica natural

> *grade*: F < D < C < B < A

> *parent_education_level*: High School < Bachelor < Master

> *study_environment*: Noisy < Moderate < Quiet

Dessa forma, vamos fazer um mapeamento ordinal manual, garantindo que um futuro algoritmo entenda os pesos corretos de cada categoria

```python
# Criamos dicionários e colocamos o peso de cada categoria
mapa_educacao = {'High School': 0, 'Bachelor': 1, 'Master': 2}
mapa_ruido = {'Noisy': 0, 'Moderate': 1, 'Quiet': 2}
mapa_notas = {'F': 0, 'D': 1, 'C': 2, 'B': 3, 'A': 4}

# Aplicamos o mapeamento em cada coluna
df['parent_education_level'] = df['parent_education_level'].map(mapa_educacao)
df['study_environment'] = df['study_environment'].map(mapa_ruido)
df['grade'] = df['grade'].map(mapa_notas)

# Resultado final e os novos tipos de dados
print(df.head())
print("\n--- Novos tipos de dados ---")
print(df.info())
```

### Padronização

Diferentes colunas apresentam diferentes escalas de números. Por exemplo, notas anteriores (previous_score) varia de $0$ a $100$, enquanto que horas de sono (sleep_hours) varia de $4$ a $9$. O futuro algoritmo, na hora de prever alguma variável, não sabe o que é sono ou nota, ele só enxerga os números absolutos. Como as notas operam na casa dos $100$ e o sono na casa dos $10$, o algoritmo vai achar que as notas são muito "maiores" e mais importantes que as horas de sono. Dessa forma, é necessário que **padronizemos as réguas**, ou seja, nós *esprememos* ou *esticamos* todas as colunas numéricas para que elas tenham a mesma importância.


**A Matemática da Padronização (Z-Score)**:

A fórmula matemática para transformar cada número antigo no novo número padronizado é:$$z = \frac{x - \mu}{\sigma}$$

* $x$: é o valor original do aluno (ex: $8$ horas de sono).
* $\mu$ (Média): é a média de horas de sono de todos os $1000$ alunos.
* $\sigma$ (Desvio Padrão): é uma medida de o quanto os alunos "variam" em torno da média.

> **A parte de cima $(x - \mu)$:** ao subtrair a média, nós "centralizamos" os dados no zero. Se um aluno dorme exatamente a média da turma, a conta dá zero. Se ele dorme mais que a média, o número fica positivo. Se dorme menos, fica negativo.

>  **A divisão por $\sigma$:** ao dividir pelo desvio padrão, nós garantimos que a dispersão dos dados fique igual para todas as colunas.

Depois de aplicar essa fórmula, a média de todas as colunas passará a ser exatamente $0$, e os valores vão variar, em sua maioria, entre $-3$ e $+3$. A relação entre quem vai melhor ou pior continua intacta, só mudamos a "escala de medida".

```python
# Importando a ferramenta de padronização
from sklearn.preprocessing import StandardScaler

# Criamos o escalonador
scaler = StandardScaler()

# Listamos apenas as colunas numéricas contínuas que possuem escalas muito diferentes
colunas_para_escalonar = [
    'final_score', 'previous_score', 'math_prev_score', 'science_prev_score',
    'language_prev_score', 'daily_study_hours', 'attendance_percentage',
    'homework_completion_rate', 'sleep_hours', 'screen_time_hours',
    'physical_activity_minutes', 'motivation_score', 'exam_anxiety_score'
]

# Aplicamos a matemática e substituímos os valores antigos pelos novos
df[colunas_para_escalonar] = scaler.fit_transform(df[colunas_para_escalonar])

# Mostrando o resultado das 5 primeiras linhas
df.head()
```
